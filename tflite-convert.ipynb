### Convert trained model to tflite for mobile application

```python
#import
import tensorflow as tf
```

```python
# Load model
model = tf.keras.models.load_model('./models/mobilenetv2_plant_disease.h5')
```

```python
model.summary()
```

```python
# Save model - using export() for SavedModel format instead of save()
model.export("./models/saved_model")  # Use export() for SavedModel format
# Alternatively: model.save("saved_model_pd.keras")  # Use .keras extension with save()
```

```python
# Convert to TFLite with float16 quantization
converter = tf.lite.TFLiteConverter.from_saved_model("./models/saved_model")
converter.optimizations = [tf.lite.Optimize.DEFAULT]
converter.target_spec.supported_types = [tf.float16]


# Convert the model with keras model
# converter = tf.lite.TFLiteConverter.from_keras_model(model)
# tflite_model = converter.convert()

tflite_model = converter.convert()

# Save TFLite model
with open('./models/model.tflite', 'wb') as f:
    f.write(tflite_model)

print("✅ Converted to TFLite (float16) successfully!")
```
